```python
from pathlib import Path

# Local workspace root for Jupyter notebooks
project_root = Path.cwd()
print(f"Project root: {project_root}")
```

```python
import os
import zipfile
from pathlib import Path


# Put the zip file next to this notebook or update this path.
zip_path = project_root / 'grocery.v9i.yolov8.zip'
extract_path = project_root / 'dataset'

if not zip_path.exists():
    raise FileNotFoundError(f'Could not find dataset zip: {zip_path}')

extract_path.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print(f'Dataset successfully extracted to {extract_path}')
```

```python
from ultralytics import YOLO

# Load the nano base model
model = YOLO('yolov8n.pt')

data_yaml = extract_path / 'data.yaml'
if not data_yaml.exists():
    raise FileNotFoundError(f'Could not find data.yaml: {data_yaml}')

# Start training using the extracted dataset config
results = model.train(
    data=str(data_yaml),
    epochs=25,
    imgsz=640,
    batch=16,
    device=0
)
```
